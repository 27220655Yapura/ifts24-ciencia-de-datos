```
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
datos = np.array([10, 20, 30, 40, 50])
print(datos)

```

```
# 1. Definición del diccionario de datos de origen
datos_institucion = {
    'estudiante': ['Ana', 'Juan', 'Carla', 'Pedro', 'Lucia'],
    'nota': [8,6,9,5,7]
}
# 2. Creación del DataFrame de Pandas (Estructura tabular bidimensional)
df_alumnos = pd.DataFrame(datos_institucion)
 
# Mostrar el DataFrame
print(df_alumnos)

# Calcular el promedio de las notas
promedio = df_alumnos['nota'].mean()
print(f"Promedio de las notas: {promedio}")

# Mostrar la nota máxima
nota_maxima = df_alumnos['nota'].max()
print(f"Nota máxima: {nota_maxima}")

# Mostrar la nota mínima

nota_minima = df_alumnos['nota'].min()
print(f"Nota mínima: {nota_minima}")
```
