```
# =====================================================================
# Ejercicio 1: Variables y operaciones básicas
# =====================================================================

def ejecutar_variables_y_operaciones():
    # Asignación de variables personales (PEP 8) [1, 2]
    nombre = "Rafa"
    edad = 37
    ciudad = "Buenos Aires"

    # Mensaje de presentación por pantalla utilizando f-strings [7]
    print(f"Hola, mi nombre es {nombre}, tengo {edad} años y vivo en {ciudad}.")

    # Asignación de variables numéricas para el procesamiento matemático [5, 6]
    a = 12
    b = 3

    # Procesamiento aritmético básico [3]
    suma = a + b
    resta = a - b
    multiplicacion = a * b

    # Salida de resultados en consola [7]
    print("\n--- Resultados de las operaciones ---")
    print(f"Suma ({a} + {b}): {suma}")
    print(f"Resta ({a} - {b}): {resta}")
    print(f"Multiplicación ({a} * {b}): {multiplicacion}")

    # Estructura condicional lógica para evitar el error ZeroDivisionError [4, 12]
    if b != 0:
        division = a / b
        print(f"División ({a} / {b}): {division}")
    else:
        print("Error: No se puede dividir por cero.")

if __name__ == "__main__":
    ejecutar_variables_y_operaciones()
```

```
# =====================================================================
# Ejercicio 2: Determinación de Mayoría de Edad
# =====================================================================

def evaluar_mayoria_edad():
    # Uso de bucle infinito controlado por break para validación [19, 20]
    while True:
        try:
            # Entrada de datos con conversión forzada a entero [14, 15]
            edad = int(input("Ingrese su edad: "))
            
            # Validación de regla de negocio
            if edad < 0:
                print("La edad no puede ser un valor negativo. Intente de nuevo.")
                continue
            break  # Entrada válida, rompemos el ciclo [19]
        except ValueError:
            # Intercepta el error de tipo en tiempo de ejecución [4, 16]
            print("Error: El dato ingresado debe ser un número entero.")

    # Estructura condicional lógica doble (if-else) [13, 18]
    if edad >= 18:
        print("Eres mayor de edad")
    else:
        print("Eres menor de edad")

if __name__ == "__main__":
    evaluar_mayoria_edad()
```

```
# =====================================================================
# Ejercicio 3: Procesamiento Estadístico de Listas
# =====================================================================

def procesar_lista_y_promediar():
    # Creación de una lista de 5 números (PEP 8) [22, 27]
    lista_numeros = [10.5, 25.0, 45.5, 20.0, 49.0]
    
    # Inicialización del acumulador
    suma_total = 0.0

    print("--- Recorrido secuencial de la lista ---")

    # Bucle definido for para recorrer colecciones ordenadas [24, 28]
    for numero in lista_numeros:
        print(f"Elemento de la lista: {numero}")
        suma_total += numero  # Operador de asignación incrementada (suma_total = suma_total + numero) [30]

    # Obtener dinámicamente la longitud de la lista con len() [25, 26]
    cantidad_elementos = len(lista_numeros)

    # Cálculo matemático del promedio aritmético
    promedio = suma_total / cantidad_elementos

    # Mostrar los resultados consolidados con formato de dos decimales [7]
    print("\n--- Estadísticas Consolidadas ---")
    print(f"Suma total de los números: {suma_total:.2f}")
    print(f"Promedio de los números  : {promedio:.2f}")

if __name__ == "__main__":
    procesar_lista_y_promediar()
```
